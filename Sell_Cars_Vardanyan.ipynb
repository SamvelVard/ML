```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sandeep1080/used-car-sales")

print("Path to dataset files:******************", path)
```

```python
import pandas as pd
```

```python
df = pd.read_csv(f"{path}/used_car_sales.csv")
```

Описание столбцов:

**ID:**** Уникальный идентификатор каждой записи.

**Distributor Name:** Название дистрибьютора автомобилей.

**Location:** Местоположение офиса дистрибьютора.

**Car Name:** Конкретная модель автомобиля.

**Manufacturer Name:** Название производителя автомобиля.

**Car Type:** Тип кузова (например, седан, внедорожник, хетчбэк и т. д.).

**Color:** Цвет автомобиля.

**Gearbox:** Тип коробки передач (например, механическая, автоматическая).

**Number of Seats:** Общее количество мест в автомобиле.

**Number of Doors:** Количество дверей автомобиля.

**Energy:** Тип топлива (например, бензин, дизель, электротяга).

**Manufactured Year:** Год выпуска автомобиля.

**Price-$:** Цена автомобиля в долларах США (USD).

**Mileage-KM:** Пробег автомобиля в километрах.

**Engine Power-HP:** Мощность двигателя в лошадиных силах (л.с.).

**Purchased Date:** Дата приобретения автомобиля дистрибьютором.

**Car Sale Status:** Статус продажи (продано / не продано).

**Sold Date:** Дата продажи автомобиля клиенту.

**Purchased Price-$:** Цена закупки (стоимость, уплаченная дистрибьютором).

**Sold Price-$:** Цена продажи (стоимость, уплаченная клиентом).

**Margin-%:** Размер маржи дистрибьютора в процентах.

**Sales Agent Name:** Имя торгового представителя, закрывшего сделку.

**Sales Rating:** Рейтинг торгового представителя, присвоенный дистрибьютором.

**Sales Commission-$:** Комиссионные, выплаченные дистрибьютором торговому представителю.

**Feedback:** Отзыв клиента о процессе покупки.
